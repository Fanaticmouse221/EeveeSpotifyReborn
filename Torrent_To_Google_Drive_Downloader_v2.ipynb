```
import libtorrent as lt
import time
import datetime

# 1. Modern session setup
settings = {
    'listen_interfaces': '0.0.0.0:6881',
    'alert_mask': lt.alert.category_t.all_categories,
    'enable_dht': True
}
ses = lt.session(settings)

# Your magnet link
link = "magnet:?xt=urn:btih:A92A7059079D0400212F6AA27CE47BB2A46487E6&dn=Lord.of.War.2005.UHD.BluRay.2160p.TrueHD.Atmos.7.1.DV.HEVC.REMUX-FraMeSToR&tr=udp%3A%2F%2Fopen.stealth.si%3A80%2Fannounce&tr=udp%3A%2F%2Fexodus.desync.com%3A6969%2Fannounce&tr=udp%3A%2F%2Ftracker.cyberia.is%3A6969%2Fannounce&tr=udp%3A%2F%2Ftracker.opentrackr.org%3A1337%2Fannounce&tr=udp%3A%2F%2Ftracker.torrent.eu.org%3A451%2Fannounce&tr=udp%3A%2F%2Fexplodie.org%3A6969%2Fannounce&tr=udp%3A%2F%2Ftracker.birkenwald.de%3A6969%2Fannounce&tr=udp%3A%2F%2Ftracker.moeking.me%3A6969%2Fannounce&tr=udp%3A%2F%2Fipv4.tracker.harry.lu%3A80%2Fannounce&tr=udp%3A%2F%2Foh.fuuuuuck.com%3A6969%2Fannounce&tr=udp%3A%2F%2Ft.133335.xyz%3A6969%2Fannounce&tr=udp%3A%2F%2Fthouvenin.cloud%3A6969%2Fannounce&tr=udp%3A%2F%2Fretracker01-msk-virt.corbina.net%3A80%2Fannounce&tr=udp%3A%2F%2Ftracker.srv00.com%3A6969%2Fannounce&tr=udp%3A%2F%2Ftracker.opentrackr.org%3A1337%2Fannounce&tr=http%3A%2F%2Ftracker.openbittrent.com%3A80%2Fannounce&tr=udp%3A%2F%2Fopentracker.i2p.rocks%3A6969%2Fannounce&tr=udp%3A%2F%2Ftracker.internetwarriors.net%3A1337%2Fannounce&tr=udp%3A%2F%2Ftracker.leechers-paradise.org%3A6969%2Fannounce&tr=udp%3A%2F%2Fcoppersurfer.tk%3A6969%2Fannounce&tr=udp%3A%2F%2Ftracker.zer0day.to%3A1337%2Fannounce"
print(f"Adding magnet link: {link[:60]}...")

# 2. Parse link and set parameters using dot notation and the correct flag names
params = lt.parse_magnet_uri(link)
params.save_path = '/content/drive/My Drive/Torrent/'

# THIS IS THE FIX: Use lt.torrent_flags instead
params.flags |= lt.torrent_flags.auto_managed | lt.torrent_flags.duplicate_is_error

handle = ses.add_torrent(params)

begin = time.time()
print(f"Download started at: {datetime.datetime.now()}")

print('Downloading Metadata...')
while not handle.has_metadata():
    time.sleep(1)
print('Got Metadata, Starting Torrent Download...')

print(f"Starting: {handle.name()}")

while not handle.status().is_seeding:
    s = handle.status()
    state_str = [
        'queued', 'checking', 'downloading metadata', 'downloading',
        'finished', 'seeding', 'allocating', 'checking fastresume'
    ]

    print(
        f'\r{s.progress * 100:.2f}% complete '
        f'(Down: {s.download_rate / 1000:.1f} kB/s | '
        f'Up: {s.upload_rate / 1000:.1f} kB/s | '
        f'Peers: {s.num_peers}) '
        f'{state_str[s.state]}',
        end=''
    )
    time.sleep(5)

end = time.time()
print(f"\n✅ {handle.name()} COMPLETE")

elapsed = end - begin
print(f"Elapsed Time: {int(elapsed // 60)} min : {int(elapsed % 60)} sec")
print(f"Download finished at: {datetime.datetime.now()}")
```
